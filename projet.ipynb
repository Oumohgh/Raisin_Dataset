# Projet : Classification des raisins avec le Machine Learning

## Contexte du projet

Vous disposez d'un jeu de données nommé `Raisin_Dataset.csv` contenant des caractéristiques géométriques de raisins appartenant à deux variétés : **Kecimen** et **Besni**.

Votre objectif est de développer un modèle de Machine Learning capable de prédire la variété d'un raisin à partir de ses caractéristiques numériques.

Le projet sera réalisé en plusieurs étapes, de l'exploration des données jusqu'à la sauvegarde du modèle final.

3. Affichez le contenu du DataFrame.

- Quelles sont les colonnes disponibles ?

- Affichez les principales statistiques descriptives des variables numériques. Quelles informations vous semblent importantes ?
- Affichez le nombre d'observations appartenant à chaque variété. Les deux classes sont-elles représentées de manière équilibrée ?

## Partie 4 — Création d'une fonction d'entraînement et d'évaluation

### Objectif

## Partie 6 — Optimisation du modèle SVM

### Objectif
Rechercher une meilleure configuration des hyperparamètres du modèle SVM.

### 6.1. Préparation de la recherche

**Questions guidées :**

1. Affichez les paramètres disponibles sur le modèle SVM.
2. Définissez un dictionnaire d'hyperparamètres contenant les valeurs candidates suivantes :
   - `C` : 1, 10, 100, 1000 et 3000 ;
   - `gamma` : 1, 0.1, 0.01, 0.001, 0.0001 et 0.0005 ;
   - `tol` : 0.001 et 0.0001.
3. Si vous souhaitez également tester `degree`, prévoyez une configuration utilisant un noyau polynomial et les valeurs 1, 2, 3, 4 et 5.

### 6.2. Recherche des meilleurs hyperparamètres

**Questions guidées :**

1. Utilisez `RandomizedSearchCV` pour rechercher une configuration performante du SVM.
2. Configurez la recherche avec `scoring='f1_weighted'`, `cv=5` et `n_iter=100`.
3. Lancez la recherche sur les données d'entraînement.
4. Affichez les meilleurs hyperparamètres identifiés.
5. Relevez le meilleur score F1 pondéré obtenu pendant la validation croisée.

### 6.3. Évaluation du modèle optimisé

**Questions guidées :**

1. Utilisez le meilleur estimateur obtenu après la recherche pour prédire les classes du jeu de test.
2. Affichez le rapport de classification du modèle optimisé.
3. Calculez son MCC.
4. Tracez sa courbe d'apprentissage en utilisant le bon estimateur.
5. Comparez les résultats du SVM optimisé avec ceux du SVM initial.
6. Déterminez si l'optimisation a réellement amélioré les performances du modèle.
